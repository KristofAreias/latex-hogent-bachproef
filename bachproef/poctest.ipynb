probleem met Foodfacts database: de afbeeldingen gelinked aan de producten zijn soms niet dezelfde producten. Bij dit voorbeeld is key 4 en 5 een ander product... Inconsistencies zoals dit kan het model helemaal fout zetten. Er is dus weer eens echt nood op consistent en goed opgebouwde databases. Ook zijn vele foto's te veel ingezoomed zoals op de nutrition en ingredienten tabellen. Deze zijn bijna altijd hetzelfde (behalve de waarden) en zijn dus bijna onmogelijk deftig te linken, deze worden dus niet gebruikt. 

We moetten dus zien of het model beter scoort op enkel front afbeeldingen of alle afbeeldingen meegenomen. Voor sommige producten zijn er afbeeldingen van alle kanten maar bij andere staan er foute producten bij. We moetten dus zien of dit de moeite is.

Nu we weten hoe de database opgebouwd is kunnen we de juiste informatie eruit selecteren zodat we per row de correcte url maken die het product toont. Zo kan er gezocht worden naar hetzelfde product in de shelves

```python
def build_image_url(code, key, rev, size="400"):
    code = str(code).zfill(13)

    prefix = code[:-4]
    suffix = code[-4:]

    parts = [prefix[i:i+3] for i in range(0, len(prefix), 3) if prefix[i:i+3]]
    parts.append(suffix)

    path = "/".join(parts)

    # Mogelijkheid 1: named images (front_fr.26.400.jpg)
    if rev is not None and not key.isdigit():
        url = f"https://images.openfoodfacts.org/images/products/{path}/{key}.{rev}.{size}.jpg"

    # Mogelijkheid 2: numeric images (5.400.jpg)
    if key.isdigit():
        url = f"https://images.openfoodfacts.org/images/products/{path}/{key}.{size}.jpg"
    
    print(url)
    return url

def get_front_image_url(images, code):
    if images is None:
        return None
    
    result = {}


    for img in images:
        key = img.get("key", "")
        rev = img.get("rev")  

        url = build_image_url(code, key, rev)

        if url is None:
            continue

        # store by key
        result[f"url_{key}"] = url

    return result
```

```python
df_foodfacts["image_url"] = df_foodfacts.apply(get_front_image_url, axis=1)

df_foodfacts = df_foodfacts[df_foodfacts["image_url"].notna()]
```

```python
def expand_dataframe(df):
    url_columns = []

    for _, row in df.iterrows():
        urls = get_front_image_url(row["images"], row["code"])
        url_columns.append(urls)

    url_df = pd.DataFrame(url_columns)

    # merge original + new URL columns
    df_new = pd.concat([df.reset_index(drop=True), url_df], axis=1)

    return df_new

df_enriched = expand_dataframe(df_foodfacts)
display(df_enriched)
```
