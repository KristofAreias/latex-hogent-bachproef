Gebruik maken van SKU db (usage documentation followed from: https://docs.ultralytics.com/datasets/detect/sku-110k/#dataset-yaml)

```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolov8n.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data="SKU-110K.yaml", epochs=50, imgsz=416, batch=32, device=0, workers=8)
```
